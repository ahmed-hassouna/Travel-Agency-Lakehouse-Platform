# Defining Schema

```python
bronze_aircrafts_path = "s3://travel-analytics-bronze/landing_parquet/iti/src_aircrafts/"

# Read the table
bronze_aircrafts_df = spark.read.format("parquet").load(bronze_aircrafts_path)
bronze_aircrafts_df.printSchema()
display(bronze_aircrafts_df)
```

```python
from pyspark.sql.types import (
    StructType, StructField,
    StringType, DoubleType,
    TimestampType, MapType
)

# Predefined schema for aircrafts_bronze table
aircrafts_bronze_schema = StructType([
    # Airbyte technical columns
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),

    # CDC columns
    StructField("_ab_cdc_lsn", DoubleType(), True),

    # Aircraft details
    StructField("aircraft_id", StringType(), True),
    StructField("aircraft_iata", StringType(), True),
    StructField("aircraft_icao", StringType(), True),
    StructField("aircraft_name", StringType(), True),

    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),

    # Additional Airbyte metadata
    StructField("_airbyte_additional_properties", MapType(StringType(), StringType(), True), True)
])
```

# the predefined scheama for the table

```python
aircrafts_bronze_schema_df = spark.read \
    .schema(aircrafts_bronze_schema) \
    .parquet("s3://travel-analytics-bronze/landing_parquet/iti/src_aircrafts/")
aircrafts_bronze_schema_df.display()
```

# Write aircrafts data to the Bronze Delta table (append mode)


```python
(
    aircrafts_bronze_schema_df.write\
    .format("delta")\
    .mode("append")\
    .save("s3://travel-analytics-bronze/delta/bronze/aircrafts/")
)
```

# Read aircrafts data with enforced schema and write to Bronze Delta table

```python
from pyspark.sql.types import StructType
schema = aircrafts_bronze_schema_df.schema

df = spark.read.schema(schema).parquet(bronze_aircrafts_path)


df.write.format("delta").mode("append").save(
    "s3://travel-analytics-bronze/delta/bronze/aircrafts/"
)
```
