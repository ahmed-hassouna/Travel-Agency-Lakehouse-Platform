# Defining the schema

```python
bronze_airlines_path = "s3://travel-analytics-bronze/landing_parquet/iti/src_airlines/"

# Read the table
bronze_airlines_df = spark.read.format("parquet").load(bronze_airlines_path)
bronze_airlines_df.printSchema()
display(bronze_airlines_df)
```

#  the predefined scheama for the table

```python
from pyspark.sql.types import (
    StructType, StructField,
    StringType, LongType, DoubleType,
    TimestampType, MapType
)

# Predefined schema for airlines_bronze table
airlines_bronze_schema = StructType([
    # Airbyte technical columns
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),
    
    # Airline basic info
    StructField("alias", StringType(), True),
    StructField("country", StringType(), True),
    StructField("airline_id", LongType(), True),
    StructField("fleet_size", LongType(), True),

    # CDC columns
    StructField("_ab_cdc_lsn", DoubleType(), True),
    
    # Airline codes and names
    StructField("airline_iata", StringType(), True),
    StructField("airline_icao", StringType(), True),
    StructField("airline_name", StringType(), True),

    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),

    # Additional Airbyte properties
    StructField("_airbyte_additional_properties", MapType(StringType(), StringType(), True), True)
])
```

```python
airlines_bronze_schema_df = spark.read \
    .schema(airlines_bronze_schema) \
    .parquet("s3://travel-analytics-bronze/landing_parquet/iti/src_airlines/")
airlines_bronze_schema_df.display()
```

```python
(
    airlines_bronze_schema_df.write\
    .format("delta")\
    .mode("append")\
    .save("s3://travel-analytics-bronze/delta/bronze/airlines/")
)
```

# Read airlines data with enforced schema and write to Bronze Delta table

```python
from pyspark.sql.types import StructType
schema = airlines_bronze_schema_df.schema

df = spark.read.schema(schema).parquet(bronze_airlines_path)


df.write.format("delta").mode("append").save(
    "s3://travel-analytics-bronze/delta/bronze/airlines/"
)
```
