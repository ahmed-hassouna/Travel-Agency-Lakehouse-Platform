# Defining Schema

```python
bronze_airports_path = "s3://travel-analytics-bronze/landing_parquet/iti/src_airports/"

# Read the table
bronze_airports_df = spark.read.format("parquet").load(bronze_airports_path)
bronze_airports_df.printSchema()
display(bronze_airports_df)
```

# Predefine Schema 

```python
from pyspark.sql.types import (
    StructType, StructField,
    StringType, LongType, DoubleType,
    TimestampType, MapType
)

# Predefined schema for airports_bronze table
airports_bronze_schema = StructType([
    # Airbyte technical columns
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),

    # Airport location and coordinates
    StructField("city", StringType(), True),
    StructField("latitude", DoubleType(), True),
    StructField("longitude", DoubleType(), True),

    # Airport identifiers
    StructField("airport_id", LongType(), True),

    # CDC column
    StructField("_ab_cdc_lsn", DoubleType(), True),

    # Airport metadata
    StructField("airport_name", StringType(), True),
    StructField("airport_type", StringType(), True),

    # CDC metadata
    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),

    # Additional Airbyte metadata
    StructField("_airbyte_additional_properties", MapType(StringType(), StringType(), True), True)
])
```

# the predefined scheama for the table

```python
airports_bronze_schema_df = spark.read \
    .schema(airports_bronze_schema) \
    .parquet("s3://travel-analytics-bronze/landing_parquet/iti/src_airports/")
airports_bronze_schema_df.display()
```

# Write airports data to the Bronze Delta table (append mode)

```python
(
    airports_bronze_schema_df.write\
    .format("delta")\
    .mode("append")\
    .save("s3://travel-analytics-bronze/delta/bronze/airports/")
)
```

# Read airports data with enforced schema and write to Bronze Delta table

```python
from pyspark.sql.types import StructType
schema = airports_bronze_schema_df.schema

df = spark.read.schema(schema).parquet(bronze_airports_path)


df.write.format("delta").mode("overwrite").save(
    "s3://travel-analytics-bronze/delta/bronze/airports/"
)
```
