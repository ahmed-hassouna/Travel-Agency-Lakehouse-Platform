# Defining the schema

```python
bronze_customers_path = "s3://travel-analytics-bronze/landing_parquet/iti/src_customers/"

# Read the table
bronze_customers_df = spark.read.format("parquet").load(bronze_customers_path)
bronze_customers_df.printSchema()
display(bronze_customers_df)
```

#  the predefined scheama 

```python
from pyspark.sql.types import (
    StructType, StructField, StringType, LongType, DoubleType,
    BooleanType, TimestampType, DateType, MapType, StructType
)

customers_bronze_schema = StructType([

    # Airbyte technical columns
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),

    StructField(
        "_airbyte_additional_properties",
        MapType(StringType(), StringType(), True),
        True
    ),

    # CDC columns
    StructField("_ab_cdc_lsn", DoubleType(), True),
    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),

    # Customer main columns
    StructField("id", LongType(), True),
    StructField("email", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("country", StringType(), True),

    # birth_date is a struct → you MUST define it as StructType
    StructField(
        "birth_date",
        StructType([
            StructField("member0", DateType(), True),
            StructField("member1", StringType(), True)
        ]),
        True
    ),

    StructField("first_name", StringType(), True),
    StructField("family_name", StringType(), True),
    StructField("phone_number", StringType(), True),
])
```

```python
customers_bronze_schema_df = spark.read \
    .schema(customers_bronze_schema) \
    .parquet("s3://travel-analytics-bronze/landing_parquet/iti/src_customers/")
customers_bronze_schema_df.display()
```

# Write data

```python
(
    customers_bronze_schema_df.write\
    .format("delta")\
    .mode("append")\
    .save("s3://travel-analytics-bronze/delta/bronze/customers/")
)
```

# Read customers data with enforced schema and write to Bronze Delta table

```python
from pyspark.sql.types import StructType
schema = customers_bronze_schema_df.schema

df = spark.read.schema(schema).parquet(bronze_customers_path)


df.write.format("delta").mode("overwrite").save(
    "s3://travel-analytics-bronze/delta/bronze/customers/"
)
```
