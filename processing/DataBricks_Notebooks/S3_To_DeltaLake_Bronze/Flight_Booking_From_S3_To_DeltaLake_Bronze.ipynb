## Defining the schema

```python
from pyspark.sql.types import (
    StructType, StructField, StringType, LongType, DoubleType,
    BooleanType, TimestampType, DateType, MapType
)

flight_bookings_schema = StructType([
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),

    StructField("price", LongType(), True),
    StructField("trip_id", LongType(), True),
    StructField("airline_id", LongType(), True),

    StructField("_ab_cdc_lsn", DoubleType(), True),

    StructField("aircraft_id", StringType(), True),
    StructField("airport_dst", LongType(), True),
    StructField("airport_src", LongType(), True),
    StructField("customer_id", LongType(), True),
    StructField("seat_number", StringType(), True),

    StructField("arrival_date", StructType([
        StructField("member0", DateType(), True),
        StructField("member1", StringType(), True)
    ]), True),

    StructField("arrival_time", StringType(), True),

    StructField("booking_date", StructType([
        StructField("member0", DateType(), True),
        StructField("member1", StringType(), True)
    ]), True),

    StructField("booking_time", StringType(), True),

    StructField("travel_class", StringType(), True),
    StructField("flight_number", StringType(), True),
    StructField("booking_status", StringType(), True),

    StructField("departure_date", StructType([
        StructField("member0", DateType(), True),
        StructField("member1", StringType(), True)
    ]), True),

    StructField("departure_time", StringType(), True),
    StructField("payment_method", StringType(), True),

    StructField("discount_amount", DoubleType(), True),
    StructField("flight_duration", StringType(), True),

    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),

    StructField("final_ticket_price", DoubleType(), True),

    StructField("_airbyte_additional_properties", MapType(StringType(), StringType()), True)
])
flight_bookings_bronze_df = spark.read.schema(flight_bookings_schema).parquet(
    "s3://travel-analytics-bronze/landing_parquet/iti/src_flight_bookings/"
)

display(flight_bookings_bronze_df)
```

## Casting dates and writting the data to the delta lake using the append mode

```python
from pyspark.sql.functions import col, to_date
flight_bookings_bronze_clean = (
    flight_bookings_bronze_df
        .withColumn("arrival_date", to_date(col("arrival_date.member0")))
        .withColumn("booking_date", to_date(col("booking_date.member0")))
        .withColumn("departure_date", to_date(col("departure_date.member0")))
)

(
flight_bookings_bronze_clean.write
    .format("delta")
    .mode("append")
    .save("s3://travel-analytics-bronze/delta/bronze/flight_bookings/")
)
```
