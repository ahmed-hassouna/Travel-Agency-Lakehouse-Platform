## copy the schema from printschema to chatgpt to suggest u the predefined scheama for the table

```python
from pyspark.sql.types import (
    StructType, StructField, StringType, LongType, DoubleType,
    BooleanType, TimestampType, DateType, MapType
)

# Predefined schema for hotels_bronze table
hotels_bronze_schema = StructType([
    # Airbyte technical columns
    StructField("_airbyte_ab_id", StringType(), True),
    StructField("_airbyte_emitted_at", TimestampType(), True),
    StructField("_airbyte_additional_properties", MapType(StringType(), StringType(), True), True),
    
    # CDC columns
    StructField("_ab_cdc_lsn", DoubleType(), True),
    StructField("_ab_cdc_deleted_at", StringType(), True),
    StructField("_ab_cdc_updated_at", StringType(), True),
    
    # Hotel identifiers and basic info
    StructField("hotel_id", LongType(), True),
    StructField("hotel_name", StringType(), True),
    StructField("hotel_address", StringType(), True),
    
    # Location information
    StructField("city", StringType(), True),
    StructField("country", StringType(), True),
    
    # Hotel specifications
    StructField("room_count", LongType(), True),
    StructField("star_rating", DoubleType(), True),
    StructField("hotel_score", DoubleType(), True)
])
```

```python
hotels_bronze_schema_df = spark.read \
    .schema(hotels_bronze_schema) \
    .parquet("s3://travel-analytics-bronze/landing_parquet/iti/src_hotels/")
hotels_bronze_schema_df.display()
```

```python
(
hotels_bronze_schema_df.write\
    .format("delta")\
    .mode("append")\
    .save("s3://travel-analytics-bronze/delta/bronze/hotels/")
)
```
