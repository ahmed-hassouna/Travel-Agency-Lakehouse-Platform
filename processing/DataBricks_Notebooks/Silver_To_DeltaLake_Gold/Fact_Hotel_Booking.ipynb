## loading Fact_Hotel_Booking Related Data from DeltaLake

```python
# ============================================================================
# CONFIGURATION - Update these paths as needed
# ============================================================================

S3_PATHS = {
    "hotel_bookings_silver": "s3://travel-analytics-bronze/delta/silver/hotel_bookings/",
    "dim_hotel": "s3://travel-analytics-bronze/delta/gold/Dim_Hotel/",
    "dim_date": "s3://travel-analytics-bronze/delta/gold/Dim_Date/",
    "dim_time": "s3://travel-analytics-bronze/delta/gold/Dim_Time/",
    "dim_customer": "s3://travel-analytics-bronze/delta/gold/Dim_Customer/", 
    "fact_hotel_booking_output": "s3://travel-analytics-bronze/delta/gold/Fact_Hotel_Booking/"
}

# ============================================================================
# STEP 1: Read Data from Delta Lake
# ============================================================================

print("="*80)
print("READING DATA FROM DELTA LAKE")
print("="*80)

# Read Hotel Bookings (Silver layer)
hotel_bookings_silver_df = spark.read.format("delta").load(S3_PATHS["hotel_bookings_silver"])
print(f"Hotel Bookings Silver: {hotel_bookings_silver_df.count():,} records")

# Reading Hotel Dimension (Gold layer)
dim_hotel_df = spark.read.format("delta").load(S3_PATHS["dim_hotel"])
print(f"Dim_Hotel: {dim_hotel_df.count():,} records")

# Reading Hotel Dimension (Gold layer)
date_df = spark.read.format("delta").load(S3_PATHS["dim_date"])
print(f"Dim_Date: {date_df.count():,} records")

# Reading Time Dimension (Gold layer)
time_df = spark.read.format("delta").load(S3_PATHS["dim_time"])
print(f"Dim_Time: {time_df.count():,} records")

#Reading Customer Dimension (Gold layer)
customer_df = spark.read.format("delta").load(S3_PATHS["dim_customer"])
print(f"Dim_Customer: {customer_df.count():,} records")
```

## Fact_Hotel_Booking Table


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# ========================================================
# STEP 0: Base fact table
# ========================================================
fact_df = hotel_bookings_silver_df

# ========================================================
# STEP 1: Join Hotel Dimension
# ========================================================
print("Joining Hotel Dimension...")
fact_df = fact_df.join(
    dim_hotel_df.select(
        F.col("Dim_Hotel_SK"),
        F.col("Hotel_ID_BK")
    ),
    fact_df.Hotel_Id == F.col("Hotel_ID_BK"),
    "left"
)

# ========================================================
# STEP 2: Join Customer Dimension
# ========================================================
print("Joining Customer Dimension...")
fact_df = fact_df.join(
    customer_df.select(
        F.col("Dim_Customer_SK"),
        F.col("Customer_ID_BK")
    ),
    fact_df.Customer_Id == F.col("Customer_ID_BK"),
    "left"
)

# ========================================================
# STEP 3: Join Check-In Date Dimension
# ========================================================
print("Joining Check-In Date Dimension...")
date_checkin = date_df.select(
    F.col("Dim_Date_SK").alias("CheckIn_SK"),
    F.col("Full_Date").alias("CheckIn_Full_Date")
)
fact_df = fact_df.join(
    date_checkin,
    fact_df.Check_In_Date == F.col("CheckIn_Full_Date"),
    "left"
).drop("CheckIn_Full_Date")

# ========================================================
# STEP 4: Join Check-Out Date Dimension
# ========================================================
print("Joining Check-Out Date Dimension...")
date_checkout = date_df.select(
    F.col("Dim_Date_SK").alias("CheckOut_SK"),
    F.col("Full_Date").alias("CheckOut_Full_Date")
)
fact_df = fact_df.join(
    date_checkout,
    fact_df.Check_Out_Date == F.col("CheckOut_Full_Date"),
    "left"
).drop("CheckOut_Full_Date")

# ========================================================
# STEP 5: Join Booking Date Dimension
# ========================================================
print("Joining Booking Date Dimension...")
date_booking = date_df.select(
    F.col("Dim_Date_SK").alias("BookingDate_SK"),
    F.col("Full_Date").alias("Booking_Full_Date")
)
fact_df = fact_df.join(
    date_booking,
    fact_df.Booking_Date == F.col("Booking_Full_Date"),
    "left"
).drop("Booking_Full_Date")

# ========================================================
# STEP 6: Join Booking Time Dimension
# ========================================================
print("Joining Booking Time Dimension...")
time_booking = time_df.select(
    F.col("Time_Dim_SK").alias("BookingTime_SK"),
    F.col("Full_Time").alias("Booking_Full_Time")
)
fact_df = fact_df.join(
    time_booking,
    fact_df.Booking_Time == F.col("Booking_Full_Time"),
    "left"
).drop("Booking_Full_Time")

# ========================================================
# STEP 7: Drop duplicate columns if exist
# ========================================================
cols = fact_df.columns
seen = set()
final_cols = []
for c in cols:
    if c not in seen:
        final_cols.append(c)
        seen.add(c)
fact_df = fact_df.select(*final_cols)

# ========================================================
# STEP 8: Generate Surrogate Key
# ========================================================
print("Generating Hotel_Booking_SK...")
window_spec = Window.orderBy(F.monotonically_increasing_id())
fact_df = fact_df.withColumn(
    "Hotel_Booking_SK",
    F.row_number().over(window_spec)
)

# ========================================================
# STEP 9: Select Final Columns
# ========================================================
print("Selecting final columns...")
fact_hotel_booking = fact_df.select(
    # Primary Key
    F.col("Hotel_Booking_SK"),

    # Business Keys
    F.col("Hotel_Id").alias("Hotel_ID_BK"),
    F.col("Customer_Id").alias("Customer_ID_BK"),

    # Foreign Keys
    F.col("Dim_Hotel_SK").alias("Dim_Hotel_FK"),
    F.col("Dim_Customer_SK").alias("Dim_Customer_FK"),
    F.col("CheckIn_SK").alias("Check_In_Date_FK"),
    F.col("CheckOut_SK").alias("Check_Out_Date_FK"),
    F.col("BookingDate_SK").alias("Booking_Date_FK"),
    F.col("BookingTime_SK").alias("Booking_Time_FK"),

    # Measures
    F.col("Stay_Duration_Days").alias("Number_Of_Nights"),
    F.col("Price_Per_Night"),
    F.col("Total_Price"),

    # Degenerate Dimensions
    F.col("Payment_Method"),
    F.col("Payment_Status"),
    F.col("Breakfast_Included"),

    # Additional Attributes
    F.col("Booking_Lead_Time_Days")
)

# ========================================================
# STEP 10: Validate
# ========================================================
fact_hotel_booking.printSchema()
fact_hotel_booking.show(5)
```
