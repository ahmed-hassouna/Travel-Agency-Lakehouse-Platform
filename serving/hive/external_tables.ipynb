## Create external tables in Hive Metastore pointing to Delta Lake files

```python
# ============================================================================
# CREATE TABLES OVER DELTA LAKE FILES
# ============================================================================

print("="*80)
print("CREATING TABLES OVER DELTA LAKE FILES")
print("="*80)

# Database name (optional - you can create tables in default database)
database_name = "TravelAgent_LakeHouse"

# Create database if it doesn't exist
spark.sql(f"CREATE DATABASE IF NOT EXISTS {database_name}")
print(f" Database '{database_name}' created/verified")

# ============================================================================
# DIMENSION TABLES
# ============================================================================

# 1. Dim_Aircraft
print("\n1. Creating Dim_Aircraft table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Aircraft
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Aircraft/'
""")
print("✓ Dim_Aircraft table created")

# 2. Dim_Airline
print("\n2. Creating Dim_Airline table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Airline
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Airline/'
""")
print("✓ Dim_Airline table created")

# 3. Dim_Airport
print("\n3. Creating Dim_Airport table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Airport
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Airport/'
""")
print("✓ Dim_Airport table created")

# 4. Dim_Customer
print("\n4. Creating Dim_Customer table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Customer
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Customer/'
""")
print("✓ Dim_Customer table created")

# 5. Dim_Date
print("\n5. Creating Dim_Date table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Date
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Date/'
""")
print("✓ Dim_Date table created")

# 6. Dim_Hotel
print("\n6. Creating Dim_Hotel table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Hotel
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Hotel/'
""")
print("✓ Dim_Hotel table created")

# 7. Dim_Time
print("\n7. Creating Dim_Time table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Dim_Time
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Dim_Time/'
""")
print("✓ Dim_Time table created")

# ============================================================================
# FACT TABLES
# ============================================================================

# 8. Fact_Flight_Booking
print("\n8. Creating Fact_Flight_Booking table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Fact_Flight_Booking
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Fact_Flight_Booking/'
""")
print("✓ Fact_Flight_Booking table created")

# 9. Fact_Hotel_Booking
print("\n9. Creating Fact_Hotel_Booking table...")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {database_name}.Fact_Hotel_Booking
USING DELTA
LOCATION 's3://travel-analytics-bronze/delta/gold/Fact_Hotel_Booking/'
""")
print("✓ Fact_Hotel_Booking table created")

print("\n" + "="*80)
print("ALL TABLES CREATED SUCCESSFULLY")
print("="*80)

# ============================================================================
# VERIFY TABLES
# ============================================================================

print("\n" + "="*80)
print("VERIFYING TABLES")
print("="*80)

# Show all tables in the database
tables_df = spark.sql(f"SHOW TABLES IN {database_name}")
tables_df.show(truncate=False)

# Show record counts for each table
print("\n" + "="*80)
print("TABLE RECORD COUNTS")
print("="*80)

tables = [
    "Dim_Aircraft",
    "Dim_Airline", 
    "Dim_Airport",
    "Dim_Customer",
    "Dim_Date",
    "Dim_Hotel",
    "Dim_Time",
    "Fact_Flight_Booking",
    "Fact_Hotel_Booking"
]

for table in tables:
    try:
        count = spark.sql(f"SELECT COUNT(*) as count FROM {database_name}.{table}").collect()[0]['count']
        print(f"{table:30} : {count:,} records")
    except Exception as e:
        print(f"{table:30} : ERROR - {str(e)}")

print("\n" + "="*80)
```
